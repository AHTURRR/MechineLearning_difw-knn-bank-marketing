# Tahap 08: Comparative Analysis, Statistical Significance & Ablation Study

Notebook ini menjalankan analisis perbandingan komprehensif dan studi ablasi di atas prediksi terkunci **Hold-out Test Set (`N = 9,043`)** dari Tahap 07:
1. **Studi Ablasi Komponen (`E0` → `E5`)** dan kontribusi marjinal tiap komponen.
2. **Uji DeLong** untuk membandingkan `ROC-AUC` terkorelasi.
3. **Uji McNemar** untuk membandingkan keputusan biner `0/1`.
4. **Paired Bootstrap Hypothesis Test (`B = 1,000`)** dengan koreksi **Holm-Bonferroni** untuk `PR-AUC`, `ROC-AUC`, `F1`, `Recall`, `MCC`, dan `Mean Cost (5:1)`.
5. **Analisis Transisi Prediksi (`E0 -> E5`) & Subgrup Domain**.

```python
import importlib.util
from pathlib import Path
from IPython.display import display, Image

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
spec = importlib.util.spec_from_file_location(
    'stage08', PROJECT_ROOT / 'src' / '08_comparative_analysis_ablation.py'
)
stage08 = importlib.util.module_from_spec(spec)
spec.loader.exec_module(stage08)

res = stage08.run_stage_08()
print(f"Evaluated on N = {res['n_test']:,} Hold-out Test observations.")
```

```python
print('=== 1. Ablation Summary (E0 - E5) ===')
display(res['ablation_summary'])

print('=== 2. Marginal Component Deltas ===')
display(res['ablation_deltas'])
```

```python
print('=== 3. DeLong ROC-AUC Tests (Holm-Bonferroni) ===')
display(res['delong_df'])

print('=== 4. McNemar Tests (Holm-Bonferroni) ===')
display(res['mcnemar_df'][['comparison', 'n10_a_right_b_wrong', 'n01_a_wrong_b_right', 'chi2_cc', 'p_value_holm_chi2', 'significant_0_05']])

print('=== 5. Paired Bootstrap Tests (E5 vs E0, E1, E2, E3, E4) ===')
display(res['boot_df'][res['boot_df']['baseline_id'].isin(['E0', 'E1', 'E2', 'E3', 'E4'])])
```

```python
print('=== 6. Confusion Matrix Transitions (E0 -> E5) ===')
display(res['trans_df'])

print('=== 7. Subgroup Performance (previously_contacted & poutcome) ===')
display(res['subgroup_df'])
```
