# Tahap 09: Explainability with LIME & Kesimpulan Akhir Penelitian

Notebook ini menjalankan interpretabilitas lokal berbasis **LIME (`LimeTabularExplainer`)** pada model final terkunci **DIFW-KNN (`E5`)** di atas **Hold-out Test Set (`N = 9,043`)**:
1. Verifikasi reproduksi probabilitas `E5` terhadap Tahap 07.
2. Penjelasan lokal LIME pada 8 kasus representatif (`TP`, `FP`, `FN`, `TN`, termasuk kasus yang berhasil diselamatkan `E5` dari `FN` pada `E0`).
3. Pengukuran keselarasan (*alignment*) antara **Global Triplet Feature Weights** dan rata-rata atribusi lokal `|LIME|` pada 40 sampel terstratifikasi.
4. Visualisasi grafik LIME dan kesimpulan akhir penelitian (RM1–RM3, H1–H3).

```python
import importlib.util
from pathlib import Path
from IPython.display import display, Image

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
spec = importlib.util.spec_from_file_location(
    'stage09', PROJECT_ROOT / 'src' / '09_explainability_lime_conclusions.py'
)
stage09 = importlib.util.module_from_spec(spec)
spec.loader.exec_module(stage09)

res = stage09.run_stage_09()
print(f"E5 max reproduction diff: {res['max_repro_diff']:.3e}")
print(f"40-sample mean LIME R2 fidelity: {res['mean_fidelity']:.4f}")
print(f"Global-vs-Local Spearman rho: {res['spearman_rho']:.4f} (p = {res['spearman_p']:.4e})")
```

```python
print('=== 1. Representative Hold-out Test Cases (LIME) ===')
display(res['summary_df'])

print('=== 2. Global Triplet Weights vs Mean Local |LIME| ===')
display(res['align_df'])
```
